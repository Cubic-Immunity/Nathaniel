```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def make_sine_wave(x, A, w):
    return A * np.sin(w * x)
```

```python
x = np.linspace(0, 2*np.pi, 1000)
amplitude = [0.5, 1, 1.5, 2, 2.5]
frequency = [1, 2, 3, 4, 5]

fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(25, 5))

for A, w in zip(amplitude, frequency):
    # First subplot: Vary only amplitude (frequency = 1)
    y1 = make_sine_wave(x, A, 1)
    ax1.plot(x, y1, label=f'A = {A}')
    
    # Second subplot: Vary only frequency (amplitude = 1)
    y2 = make_sine_wave(x, 1, w)
    ax2.plot(x, y2, label=f'ω = {w}')
    
    # Third subplot: Vary both amplitude and frequency
    y3 = make_sine_wave(x, A, w)
    ax3.plot(x, y3, label=f'A = {A}, ω = {w}')

# First subplot
ax1.set_title('Varying Amplitude (w = 1)', fontsize=18)
ax1.set_xlabel('x (radians)', fontsize=16)
ax1.set_ylabel('y = A · sin(x)', fontsize=16)
ax1.legend(fontsize=14)
ax1.grid(alpha=0.5)
ax1.tick_params(labelsize=15)

# Configure second subplot
ax2.set_title('Varying Frequency (A = 1)', fontsize=18)
ax2.set_xlabel('x (radians)', fontsize=16)
ax2.set_ylabel('y = sin(wx)', fontsize=16)
ax2.legend(fontsize=14)
ax2.grid(alpha=0.5)
ax2.tick_params(labelsize=15)

# Configure third subplot
ax3.set_title('Varying Both Amplitude and Frequency', fontsize=18)
ax3.set_xlabel('x (radians)', fontsize=16)
ax3.set_ylabel('y = A · sin(wx)', fontsize=16)
ax3.legend(fontsize=14)
ax3.grid(alpha=0.5)
ax3.tick_params(labelsize=15)

plt.tight_layout()
plt.show()
```
