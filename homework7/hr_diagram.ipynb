```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.loadtxt("../../course_assignments/homework7/star_data.txt", skiprows=1)
```

```python
data.shape
print(f"There are {data.shape[0]} rows")
```

```python
magnitude = data[:, 0]
temperature = data[:, 1]

mag_min = np.min(magnitude)
mag_max = np.amax(magnitude)
mag_min_idx = np.argmin(magnitude)
mag_max_idx = np.argmax(magnitude)

temp_min = np.min(temperature)
temp_max = np.max(temperature)
temp_min_idx = np.argmin(temperature)
temp_max_idx = np.argmax(temperature)

print(f"Minimum magnitude: {mag_min}")
print(f"- Star index: {mag_min_idx}")
print(f"\nMaximum magnitude: {mag_max}")
print(f"- Star index: {mag_max_idx}")

print(f"\nMinimum temperature: {temp_min:.2f} K")
print(f"- Star index: {temp_min_idx}")
print(f"\nMaximum temperature: {temp_max:.2f} K")
print(f"- Star index: {temp_max_idx}")

print(f"\nSince magnitudes are backwards the brightest star is {mag_min} and the dimmest star is {mag_max}")
```

```python
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)

plt.figure(figsize=(8, 6))

scatter = plt.scatter(log_temperature, log_magnitude, alpha=0.6, s=20, 
                     c=temperature, cmap='plasma', edgecolors='black', 
                     linewidth=0.5, label='Stars')

plt.gca().invert_xaxis()
plt.gca().invert_yaxis()

# Brightest star (smallest magnitude)
plt.scatter(log_temperature[mag_min_idx], log_magnitude[mag_min_idx], 
           color='black', marker='*', s=300, edgecolors='white', 
           linewidth=1.5, label='Brightest Star', zorder=5)

# Dimmest star (largest magnitude)
plt.scatter(log_temperature[mag_max_idx], log_magnitude[mag_max_idx], 
           color='purple', marker='*', s=300, edgecolors='white', 
           linewidth=1.5, label='Dimmest Star', zorder=5)

# Hottest star (maximum temperature)
plt.scatter(log_temperature[temp_max_idx], log_magnitude[temp_max_idx], 
           color='green', marker='*', s=300, edgecolors='white', 
           linewidth=1.5, label='Hottest Star', zorder=5)

# Coldest star (minimum temperature)
plt.scatter(log_temperature[temp_min_idx], log_magnitude[temp_min_idx], 
           color='orange', marker='*', s=300, edgecolors='white', 
           linewidth=1.5, label='Coldest Star', zorder=5)

# Add colorbar for temperature
cbar = plt.colorbar(scatter, label='Temperature (K)')
cbar.ax.tick_params(labelsize=10)

plt.legend(fontsize=10, loc='best', framealpha=0.9)
plt.title('Hertzsprung-Russell Diagram', fontsize=16)
plt.xlabel('Temperature (log K)', fontsize=12)
plt.ylabel('Magnitude', fontsize=12)
plt.grid(alpha=0.3, linestyle='--')
plt.tick_params(labelsize=10)
plt.tight_layout()
plt.show()
```
